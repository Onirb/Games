```python
from random import randint
from IPython.display import clear_output # Import clear_output

def HigherLowerGame(score, person1_data=None, previous_person2_index=None):
    """
    Plays one round of the Higher Lower game.

    Args:
        score (int): The current score.
        person1_data (dict, optional): The data for the first person.
                                       If None, two new people are selected.
        previous_person2_index (int, optional): The index of the person who was person2
                                                in the previous round. Used to avoid repeating
                                                the exact same pair in consecutive rounds
                                                when the guess was correct.

    Returns:
        tuple: A tuple containing the updated score, the data of the correct person
               to be used as person1 in the next round, and the index of the person2
               selected in this round. Returns None for person data and index if the game ends.
    """
    num_entries = len(person_data['Name'])
    print(higherorlower)
    # Ensure there are at least two entries to play
    if num_entries < 2:
        print("Need at least two entries in person_data to play the game.")
        return score, None, None # Return current score and None to indicate game cannot start/continue

    current_person2_index = None # Initialize index for person2 in this round

    if person1_data is None:
        # First round or previous guess was wrong, select two new distinct people
        index1 = randint(0, num_entries - 1)
        index2 = randint(0, num_entries - 1)
        while index2 == index1:
            index2 = randint(0, num_entries - 1)

        person1 = {
            'Name': person_data['Name'][index1],
            'Profession': person_data['Profession'][index1],
            'From': person_data['From'][index1],
            'Followers': person_data['Followers'][index1]
        }
        person2 = {
            'Name': person_data['Name'][index2],
            'Profession': person_data['Profession'][index2],
            'From': person_data['From'][index2],
            'Followers': person_data['Followers'][index2]
        }
        current_person2_index = index2

    else:
        # Previous guess was correct, use the correct person as person1 and select a new person2
        person1 = person1_data

        # Select a new person2, ensuring it's not the same as person1 AND not the previous person2
        index2 = randint(0, num_entries - 1)
        # Find the index of person1 in the original data to avoid selecting the same person
        try:
            index1 = person_data['Name'].index(person1['Name']) # Assumes names are unique
        except ValueError:
            # Handle case where person1's name is not found (shouldn't happen if data is consistent)
            print("Error: Could not find person1 in the data.")
            return score, None, None

        # Ensure the new person2 is different from person1 and also not the previous person2
        while index2 == index1 or index2 == previous_person2_index:
            index2 = randint(0, num_entries - 1)

        person2 = {
            'Name': person_data['Name'][index2],
            'Profession': person_data['Profession'][index2],
            'From': person_data['From'][index2],
            'Followers': person_data['Followers'][index2]
        }
        current_person2_index = index2


    # Create copies to display without followers
    person1_display = person1.copy()
    del person1_display['Followers']

    person2_display = person2.copy()
    del person2_display['Followers']

    print("Compare A:")
    display(person1_display) # Display the modified copy
    print(vs)
    print("Against B:")
    display(person2_display) # Display the modified copy

    answer = input('Who has the most followers?(A or B): ').lower()

    is_correct = False
    if (answer == 'a' and person1['Followers'] > person2['Followers']) or \
       (answer == 'b' and person2['Followers'] > person1['Followers']):
        is_correct = True

    if is_correct:
        score += 1
        print('Correct!')
        print(f'Your current score: {score}')
        # Return the updated score, the correct person's data, and the index of person2
        if answer == 'a':
            return score, person1, current_person2_index
        else:
            return score, person2, current_person2_index
    else:
        print('Wrong!')
        print(f'Your final score: {score}')
        return score, None, None # Return None for person data and index to indicate game over

# Game loop
current_score = 0
person_for_next_round = None
previous_person2_index = None # Variable to store the index of person2 from the previous round

while True:
    clear_output() # Clear the output at the beginning of each round

    current_score, person_for_next_round, previous_person2_index = HigherLowerGame(current_score, person_for_next_round, previous_person2_index)

    # Check if the game ended (either not enough data or incorrect guess)
    if person_for_next_round is None:
        # Ask to play again only if the game ended due to incorrect guess and there's data to play again
        if len(person_data['Name']) >= 2: # Ensure there's enough data for a new game
             play_again_input = input("Play again? Type 'y' or 'n': ").lower()
             if play_again_input == 'y':
                 current_score = 0 # Reset score for a new game
                 person_for_next_round = None # Start with two new people
                 previous_person2_index = None # Reset previous person2 index for a new game
                 continue # Continue the while loop for a new game
        break # Exit the loop if not enough data or user doesn't want to play again
    else:
        # If the guess was correct, the loop will continue automatically
        pass # No need to ask "Play again?" here, the loop continues with the correct person

# Final score is already printed inside the function when the game ends
```
